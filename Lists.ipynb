###### Most lists we create will be dynamic, meaning we’ll build a list and then add and remove elements from it as our program runs its course.

##### <ins> Modifying elements in a list : </ins> 
###### - The syntax for modifying an element is similar to the syntax for accessing an element in a list. To change an element, use the name of the list followed by the index of the element we want to change, and then provide the new value we want that item to have.
#### <ins> Adding Elements to a List :  <font color='red'>append() method</font> </ins>
##### We might want to add a new element to a list for many reasons.
##### <ins> Inseting Elements to a List :  <font color='red'>insert() method</font>  </ins>
##### We can add a new element at any position in our list by using the insert() method. We do this by specifying the index of the new element and the value of the new item.
##### <ins> Removing Elements to a List :  <font color='red'>del() statement</font> </ins>
##### We’ll want to remove an item or a set of items from a list. 
##### <ins> Removing Elements to a List :  <font color='red'>pop() method</font> </ins>
##### - The pop() method removes the last item in a list, but it lets you work with that item after removing it. The term pop comes from thinking of a list as a stack of items and popping one item off the top of the stack. In this analogy, the top of a stack corresponds to the end of a list. 
##### - You can actually use pop() to remove an item in a list at any position by including the index of the item you want to remove in parentheses.



```python
# Modifying elements in a list
motorcycles = ["Honda", "Yamaha", "Suzuki", "Kawasaki"]
print(motorcycles)
motorcycles[0] = "Ducati"
print(motorcycles)
```

```python
# Adding elements to a list
motorcycles = ["Honda", "Yamaha", "Suzuki", "Kawasaki"]
print(motorcycles)
motorcycles.append("Ducati")
print(motorcycles)
motorcycles.append("BMW")
print(motorcycles)
motorcycles.append("Harley-Davidson")
print(motorcycles)
```


<font color='red'>End note</font> 

###### Building lists this way is very common, because you often won’t know the data your users want to store in a program until after the program is running. To put your users in control, start by defining an empty list that will hold the users’ values. Then append each new value provided to the list you just created.

```python
motorcycles = ["Honda", "Yamaha", "Suzuki", "Kawasaki"]
print(motorcycles)
motorcycles.insert(0, "Ducati")
print(motorcycles)
```

```python
motorcycles = ["Honda", "Yamaha", "Suzuki", "Kawasaki"]
print(motorcycles)
del motorcycles[2]
print(motorcycles)
```

```python
motorcycles = ["Honda", "Yamaha", "Suzuki", "Kawasaki"]
print(motorcycles)
popped_motorcycle = motorcycles.pop()
print(motorcycles)
print("I sold my " + popped_motorcycle + ".")
```

```python
motorcycles = ["Honda", "Yamaha", "Suzuki", "Kawasaki"]
print(motorcycles)

popped_motorcycle = motorcycles.pop(0)
print("I sold my " + popped_motorcycle + ".")
```
